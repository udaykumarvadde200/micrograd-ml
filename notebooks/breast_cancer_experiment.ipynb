```python
from datasets.breast_cancer.load_data import load_dataset
from datasets.breast_cancer.preprocess import prepare_data
from micrograd.engine import Value
from experiments.breast_cancer.train import train
from micrograd.nn import MLP
```

```python
X_train, X_test, y_train, y_test = prepare_data()
```

```python


model = MLP(30, [16, 8, 1])

input_row = [Value(x) for x in X_train[0]]

output = model(input_row)

print(output)
```

```python
x = Value(2.0)
y = x.exp()

print(y)

y.backward()

print(x)
```

```python
from optimizers.sgd import SGD
from experiments.breast_cancer.train import train

optimizer = SGD(model.parameters(), lr=0.01)

train(
    model,
    X_train,
    y_train,
    optimizer,
    epochs=10
)
```

```python
correct = 0

for x, y in zip(X_test, y_test):
    inputs = [Value(feature) for feature in x]

    output = model(inputs)
    probability = output.sigmoid()

    prediction = 1 if probability.data >= 0.5 else 0

    if prediction == y:
        correct += 1

accuracy = correct / len(y_test)

print("Test Accuracy:", accuracy)
```
